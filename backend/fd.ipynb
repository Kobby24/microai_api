```python
import GEOparse, sqlite3

gpl = GEOparse.get_GEO("GPL96", destdir="./geo")   # pip install GEOparse
df = gpl.table                                      # probe-level annotation

df.to_csv("GPL96_annotation.csv", index=False)

con = sqlite3.connect("gpl96.db")
df.to_sql("gpl96", con, if_exists="replace", index=False)
con.execute("CREATE INDEX idx_probe ON gpl96(ID)")
con.close()
```
